```python
# how many unique studies vs total runs
df2['Study_ID'].nunique()

# de-duplicate to one row per study (so each study counts once, not once per run)
studies = df2.drop_duplicates(subset='Study_ID')

# now count where studies come from
studies['Location'].value_counts()
```

```python
studies['Body_Site'].value_counts()
studies['Sex'].value_counts()
```

```python
import matplotlib.pyplot as plt
studies['Location'].value_counts().plot(kind='bar', figsize=(10,5))
plt.title("Skin Microbiome Studies by Location (SKIOME)")
plt.ylabel("Number of Studies")
plt.tight_layout()
plt.show()
```

```python
# standardize capitalization first
studies['Sex_clean'] = studies['Sex'].str.strip().str.lower()

# look at what's there now
studies['Sex_clean'].value_counts()
```

```python
studies['Sex'].str.lower().value_counts()
```

```python
studies['Body_Site'].value_counts()
```

```python
# Quick filter: keep only rows that look like skin sites
skin_keywords = ['skin', 'epidermis', 'forearm', 'hip', 'groin', 'face', 'nose']
skin_only = studies[studies['Body_Site'].str.contains('|'.join(skin_keywords), case=False, na=False)]
skin_only['Body_Site'].value_counts()
```
